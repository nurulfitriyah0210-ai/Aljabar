## 2. Pengerjaan Matriks Satu Per Satu
Rumus dasar:  
$$\begin{pmatrix} a & b \\ c & d \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} x' \\ y' \end{pmatrix}$$

### A. Transformasi A (2, 3) → B (2, 1)
*   **Baris 1:** $2a + 3b = 2 \rightarrow$ Ambil $a = 1, b = 0$
*   **Baris 2:** $2c + 3d = 1 \rightarrow$ Ambil $c = 0, 3d = 1 \Rightarrow d = 1/3$
*   **Matriks:** $\begin{pmatrix} 1 & 0 \\ 0 & 1/3 \end{pmatrix}$

### B. Transformasi D (2, 4) → E (2, 0)
*   **Baris 1:** $2a + 4b = 2 \rightarrow$ Ambil $a = 1, b = 0$
*   **Baris 2:** $2c + 4d = 0 \rightarrow$ Ambil $c = 0, d = 0$
*   **Matriks:** $\begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$

### C. Transformasi E (2, 0) → F (4, 0)
*   **Baris 1:** $2a + 0b = 4 \Rightarrow 2a = 4 \Rightarrow a = 2$
*   **Baris 2:** $2c + 0d = 0 \Rightarrow c = 0$ (Gunakan $d = 1$ sebagai standar)
*   **Matriks:** $\begin{pmatrix} 2 & 0 \\ 0 & 1 \end{pmatrix}$

### D. Transformasi B (2, 1) → C (4, 1)
*   **Baris 1:** $2a + 1b = 4 \rightarrow$ Ambil $a = 2, b = 0$
*   **Baris 2:** $2c + 1d = 1 \rightarrow$ Ambil $c = 0, d = 1$
*   **Matriks:** $\begin{pmatrix} 2 & 0 \\ 0 & 1 \end{pmatrix}$

### E. Transformasi A (2, 3) → C (4, 1)
*   **Baris 1:** $2a + 3b = 4 \rightarrow$ Ambil $a = 2, b = 0$
*   **Baris 2:** $2c + 3d = 1 \rightarrow$ Ambil $c = 0, 3d = 1 \Rightarrow d = 1/3$
*   **Matriks:** $\begin{pmatrix} 2 & 0 \\ 0 & 1/3 \end{pmatrix}$

### F. Transformasi D (2, 4) → F (4, 0)
*   **Baris 1:** $2a + 4b = 4 \rightarrow$ Ambil $a = 2, b = 0$
*   **Baris 2:** $2c + 4d = 0 \rightarrow$ Ambil $c = 0, d = 0$
*   **Matriks:** $\begin{pmatrix} 2 & 0 \\ 0 & 0 \end{pmatrix}$
